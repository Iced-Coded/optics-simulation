```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

def simulate_fiber_core(angle_degrees, core_radius, length, y_offset=0):
    """Симулює промінь світла в одній жилі зі зміщенням по осі Y."""
    angle_rad = np.radians(angle_degrees)

    x_positions = [0]
    y_positions = [y_offset]

    current_x = 0
    direction_y = 1

    if angle_degrees == 0:
        return [0, length], [y_offset, y_offset]

    x_step = core_radius / np.tan(angle_rad)

    while current_x < length:
        prev_x = current_x
        prev_y = y_positions[-1]

        current_x += x_step
        current_y = y_offset + (core_radius * direction_y)

        if current_x > length:
            dx = length - prev_x
            dy = dx * np.tan(angle_rad) * direction_y
            x_positions.append(length)
            y_positions.append(prev_y + dy)
            break

        x_positions.append(current_x)
        y_positions.append(current_y)

        direction_y *= -1

    return x_positions, y_positions

# --- ІНТЕРАКТИВНА ЧАСТИНА ---

# Загортаємо створення графіка у функцію
def draw_cable(angle1, angle2, angle3, angle4, cable_length):
    core_radius = 1.2

    # Використовуємо значення з повзунків
    cores = [
        {'y_offset': 6,  'angle': angle1, 'color': 'red'},
        {'y_offset': 2,  'angle': angle2, 'color': 'green'},
        {'y_offset': -2, 'angle': angle3, 'color': 'orange'},
        {'y_offset': -6, 'angle': angle4, 'color': 'purple'}
    ]

    plt.figure(figsize=(12, 6))

    for i, core in enumerate(cores):
        y_c = core['y_offset']
        angle = core['angle']
        color = core['color']

        plt.axhline(y_c + core_radius, color='blue', linestyle='-', alpha=0.4)
        plt.axhline(y_c - core_radius, color='blue', linestyle='-', alpha=0.4)

        x, y = simulate_fiber_core(angle, core_radius, cable_length, y_offset=y_c)
        plt.plot(x, y, color=color, linewidth=1.5, label=f'Жила {i+1} (кут {angle}°)')

    outer_jacket_top = cores[0]['y_offset'] + core_radius + 1
    outer_jacket_bottom = cores[-1]['y_offset'] - core_radius - 1
    plt.axhline(outer_jacket_top, color='black', linewidth=3, label='Зовнішня оболонка')
    plt.axhline(outer_jacket_bottom, color='black', linewidth=3)

    plt.xlim(0, cable_length)
    plt.ylim(outer_jacket_bottom - 1, outer_jacket_top + 1)
    plt.title("Інтерактивна симуляція оптоволоконного кабелю")
    plt.xlabel("Довжина кабелю")
    plt.ylabel("Положення (Y)")

    plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))
    plt.grid(True, alpha=0.2)
    plt.tight_layout()
    plt.show()

# Створюємо панель з повзунками над графіком
interact(draw_cable,
         angle1=widgets.IntSlider(min=0, max=85, step=1, value=15, description='Кут жили 1:'),
         angle2=widgets.IntSlider(min=0, max=85, step=1, value=30, description='Кут жили 2:'),
         angle3=widgets.IntSlider(min=0, max=85, step=1, value=45, description='Кут жили 3:'),
         angle4=widgets.IntSlider(min=0, max=85, step=1, value=60, description='Кут жили 4:'),
         cable_length=widgets.IntSlider(min=10, max=100, step=5, value=50, description='Довжина:'))
```
